# Extraer los datos desde el CSV

```python
#Ejercicio01: Establecer librerias a utilizar
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
#Ejercicio02: Definir las cabeceras del Dataset
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]

file_path = "autos.csv"

data = pd.read_csv(file_path, names=headers, sep = ",")
```

## 1. Validación de los Datos

### Descripción de las variables

- 'symboling' : Nivel de riesgo del seguro , de menor riesgo (-3) a mayor riesgto (3)
- "normalized-losses" : Es la perdida del pago promedio del seguro del auto por año y representa la perdida por año del auto. El valor oscila entre 65 a 256
- 'make' : Marca
- "aspiration" : Como el motor de combustión interno trabaja , puede ser estandar (std) o turbo 
- 'num-of-doors' : Número de puertas
- "body-style" : Carroceria
- 'drive-wheels' : Tracción de las ruedas
- 'engine-location' : Localización del motor
- 'wheel-base' : Distancia entre ejes 
- 'length' : Longitud
- 'width' : Ancho
- 'height': Alto
- 'curb-weight' : Peso del auto con equipación completa
- 'engine-type': Tipo de motor
- 'num-of-cylinders' : Número de cilindros
- 'engine-size' : Tamaño del motor
- 'fuel-system' : Sistema de combustible
- 'bore' : Diámetro
- 'stroke' : Distancia recorrida por el pistón durante cada ciclo
- 'compression-ratio': Ratio de comprensión
- 'horsepower' : Caballos de potencia
- 'horsepower-binned' : Caballos de potencia convertidos en una variable categórica 
- 'peak-rpm' : RPM pico 
- 'city-mpg' : consumo en la ciudad en millas por galon
- 'highway-mpg' : consumo en autopista en millas por galon
- 'price' : Precio
- 'city-L/100km' :  consumo en la ciudad en litros por kilometro
- 'diesel' : variable dummy para consumo por petróleo
- 'gas': variable dummy para consumo por gasoina

```python
# Ejercicio06: Definir valor perdido de Numpy
import numpy as np
type(np.nan)
```

```python
# Ejercicio07: Reemplazar los valores perdidos por el valor de Numpy
df_replaced = data.replace("?", np.nan, inplace = False)
df_replaced.head(10)
```

```python
# Ejercicio08: Reemplazar los valores NaN por la media para el precio
avg_price = df_replaced["price"].astype('float').mean()
print(avg_price)
df_replaced['price'].replace(np.nan, avg_price, inplace=True)
```

```python
#Ejercicio09: Ordenar los datos por columna de forma descendente
df_replaced.sort_values("make", ascending=False)
```

```python
#Ejercicio10: ver los tipos de datos
df_replaced.dtypes
```

```python
# Ejercicio11: Corregir el formato del precio
df_replaced['price'] = df_replaced['price'].astype('float')
df_replaced['horsepower'] = df_replaced['horsepower'].astype('Int64')
```

```python
# Ejercicio12: Verificar la correccion del formato del precio
df_replaced.dtypes
```

```python
#Ejercicio 10: Mostrar los elementos del dataFrame cuyos precios son mayores a 5000 y son marca audi
df_replaced[(df_replaced.price > 5000) & (df_replaced.make == 'audi')]
```

```python
#Ejercicio 11: Establecer la influencia de tamaño del motor y la carroceria sobre el precio del auto
df_replaced[['engine-size','horsepower','city-mpg','price']].corr()
```

```python
#Ejercicio 12: Graficar los resultados anteriores
import matplotlib.pyplot as plt
import seaborn as sbn
%matplotlib inline
sbn.heatmap(abs(df_replaced[['engine-size','horsepower','city-mpg','price']].corr()),annot=True,fmt=".1f", cmap="Blues")
```
